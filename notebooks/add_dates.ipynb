```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta
```

```python
df = pd.read_csv('../data/processed/final_processed_data/boxing-match-fighters.csv')
df
```

```python
start_date = datetime(2010, 1, 1)
end_date   = datetime(2025, 12, 31)
```

```python
def generate_date_of_match(start, end): # start and end as it can start after 2010 and end in 2025
    delta = end - start
    random_days = np.random.randint(0, delta.days)
    return start + timedelta(days=int(random_days))

assigned_dates = {}

generated_dates = []
```

```python
for idx, row in df.iterrows():
    fighterA = row["Boxer_A"]
    fighterB = row["Boxer_B"]
    
    fighter_key = tuple(sorted([fighterA, fighterB]))
    
    while True:
        d = generate_date_of_match(start_date, end_date).date()

        # Check if either fighter already has this date
        clash = False
        for f in [fighterA, fighterB]:
            if f in assigned_dates and d in assigned_dates[f]:
                clash = True
                break
        
        if not clash:
            # Assign to both fighters
            for f in [fighterA, fighterB]:
                assigned_dates.setdefault(f, set()).add(d)
            generated_dates.append(d)
            break
```

```python
df["Fight_Date"] = generated_dates
```

```python
df.to_csv("../data/processed/final_processed_data/boxing-match-fighters.csv", index=False)
```
