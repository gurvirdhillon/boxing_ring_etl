```python
import pandas as pd
```

```python
fight_results = pd.read_csv('../data/processed/fighters.csv')
df = pd.read_csv("../data/processed/cleaned_data.csv")
```

```python
df = df.drop(columns=['Unnamed: 0'], errors='ignore')
```

```python
df
```

```python
fight_results = fight_results[fight_results['Boxer_A'] != fight_results['Boxer_B']]
```

```python
fight_results
```

```python
merged_final = pd.concat([fight_results.reset_index(drop=True), 
                          df.reset_index(drop=True)], axis=1)
```

```python
merged_final = pd.concat([fight_results.reset_index(drop=True), 
                          df.reset_index(drop=True)], axis=1)
```

```python
merged_final = merged_final[merged_final['Boxer_A'] != merged_final['Boxer_B']]
```

```python
merged_final
```

```python
merged_final = merged_final.dropna(subset=['Boxer_A', 'Boxer_B'])
```

```python
merged_final
```

```python
merged_final = merged_final.drop(columns=['Unnamed: 0'], errors='ignore')
```

```python
merged_final
```

```python
cols = ['won_A', 'won_B', 'lost_A', 'lost_B', 'kos_A']

for col in cols:
    merged_final[col] = merged_final[col].fillna(0).astype(int)
```

```python
# 1. Columns for Fighter A (end with _A or contain _A)
cols_A = [col for col in merged_final.columns if col.endswith('_A')]

# 2. Columns for Fighter B (end with _B or contain _B)
cols_B = [col for col in merged_final.columns if col.endswith('_B')]

# 3. Important fight result columns (put at end if they exist)
result_cols = [col for col in merged_final.columns if col in ['result', 'won_A', 'won_B']]

# 4. Remaining general columns (not in A, B, or result)
other_cols = [col for col in merged_final.columns if col not in cols_A + cols_B + result_cols]

# Final column order
new_order = cols_A + cols_B + other_cols + result_cols

# Apply rearrangement
merged_final = merged_final[new_order]

# Preview
merged_final.head()
```
