```python
# theres currently an issue with the weight as the cleaner boxer info shows kg whereas this dataset shows lbs
df['Weight'] = pd.to_numeric(df['Weight'], errors='coerce') * 0.45359237
```

```python
df.head()
```

```python
def assign_weight_class(weight):
    if weight < 52:
        return 'Flyweight'
    elif weight < 57:
        return 'Featherweight'
    elif weight < 63:
        return 'Lightweight'
    elif weight < 69:
        return 'Welterweight'
    elif weight < 76:
        return 'Middleweight'
    elif weight < 91:
        return 'Cruiserweight'
    else:
        return 'Heavyweight'

df['Weight_Class'] = df['Weight'].apply(assign_weight_class)
```

```python
# Merge to create valid matchups (same sex & weight class)
merge_df_matchup = df.merge(
    df,
    on=['Sex', 'Weight_Class'],
    suffixes=('_A', '_B')
)

# Remove same boxer matchups
print(merge_df_matchup.columns.tolist())
```
